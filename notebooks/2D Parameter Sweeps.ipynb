# Chemotactic Sensitivity
## Constants
* DT = 1
* N = 512
* DU = 0.2
* DV = 1
* ALPHA = 0.1
* BETA = 1
* Saturation = 4
## Results
Ranges are rough estimates  
  
* 0-0.5 : Complete diffusion  
* 0.6-0.9 : Separates into dots but sometimes dense lines remain for longer  
* 1.0 : Grows into dense dots before diffusing out from those dots and consolidating into lines; lines combine as time goes on and form vertical and horizontal patterns (likely due to the way gradient and divergence are calculated)  
* 1.1 : Density increases at points before diffusing throughout the whole systems quickly; one high-density loop remains amongst a field of medium density; at points where the line is disconnected but then combines, density is very low  
* 1.2-1.6 : Almost immediately consolidates into high-density lines that move very little, leaving a field of medium density between them; many disconnected loops are left, and unclosed loops form open sections to close themselves or connect to others; number of disconnected loops increases proportional to chemotactic sensitivity  
* 1.7-2.1 : Same behavior as 1.2-1.6, but high-density regions jitter and are even smaller; virtually no movement occurs besides the jittering instability  
* 2.2+ : Same as 1.7-2.1, but program eventually blows up
